```python
x = np.linspace(-5, 5, 101)
y = np.linspace(-5, 5, 101)

X, Y = np.meshgrid(x, y)

r1 = np.sqrt((X+1)**2 + Y**2)
r1[r1 == 0] = np.nan

r2 = np.sqrt((X-1)**2 + Y**2)
r2[r2 == 0] = np.nan

r3 = np.sqrt((X-2)**2 + (Y-2)**2)
r3[r3 == 0] = np.nan


V = ((1/r1)-(1/r2)-(1/r3))

vex = (X+1)/(r1**3) - (X-1)/(r2**3) - (X-2)/(r3**3)
vey = (Y)/(r1**3) - (Y)/(r2**3) - (Y-2)/(r3**3)

Emag = np.sqrt(vex**2 + vey**2)
ux = vex / Emag
uy = vey / Emag

plt.contourf(X, Y, V, levels=50)
plt.colorbar(label="Electric Potential")

plt.quiver(
    X[::6, ::6],
    Y[::6, ::6],
    ux[::6, ::6],
    uy[::6, ::6]
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Electric Potential and Electric Field of a dipole")
plt.axis("equal")
plt.show()
```

```python
epsilon0 = 8.854e-12
k = 1 / (4 * np.pi * epsilon0)

q1 = 1e-9      # +1 nC
q2 = -1e-9     # -1 nC

x1, y1 = -1, 0
x2, y2 = 1, 0

x = np.linspace(-5, 5, 101)
y = np.linspace(-5, 5, 101)

X, Y = np.meshgrid(x, y)

r1 = np.sqrt((X - x1)**2 + (Y - y1)**2)
r2 = np.sqrt((X - x2)**2 + (Y - y2)**2)

r1[r1 == 0] = np.nan
r2[r2 == 0] = np.nan

# electric potential

V1 = k * q1 / r1
V2 = k * q2 / r2

V = V1 + V2

Ex1 = k * q1 * (X - x1) / (r1**3)
Ey1 = k * q1 * (Y - y1) / (r1**3)

Ex2 = k * q2 * (X - x2) / (r2**3)
Ey2 = k * q2 * (Y - y2) / (r2**3)

Ex = Ex1 + Ex2
Ey = Ey1 + Ey2

Emag = np.sqrt(Ex**2 + Ey**2)

ux = Ex / Emag
uy = Ey / Emag


plt.figure(figsize=(8, 7))

plt.contourf(X, Y, V, levels=50)

plt.colorbar(label="Electric Potential (V)")

plt.quiver(
    X[::6, ::6],
    Y[::6, ::6],
    ux[::6, ::6],
    uy[::6, ::6]
)

plt.scatter(x1, y1, s=120, label="+q")
plt.scatter(x2, y2, s=120, label="-q")


plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Electric Potential and Electric Field of a Dipole")

plt.legend()
plt.axis("equal")

ax = plt.gca()
ax.set_aspect("equal", adjustable="box")

plt.xlim(-5, 5)
plt.ylim(-5, 5)

plt.show()
```

```python
epsilon0 = 8.854e-12
k = 1 / (4 * np.pi * epsilon0)

Q = 1e-9
R = 1.0

x = np.linspace(-5, 5, 201)
y = np.linspace(-5, 5, 201)

X, Y = np.meshgrid(x, y)

r = np.sqrt(X**2 + Y**2)

outside = r >= R
inside = r < R

V = np.empty_like(r)

V[outside] = k * Q / r[outside]

V[inside] = (k * Q / (2 * R)) * (
    3 - r[inside]**2 / R**2
)

E = np.empty_like(r)

E[outside] = k * Q / r[outside]**2

E[inside] = k * Q * r[inside] / R**3

Ex = np.zeros_like(r)
Ey = np.zeros_like(r)

nonzero = r != 0

Ex[nonzero] = E[nonzero] * X[nonzero] / r[nonzero]
Ey[nonzero] = E[nonzero] * Y[nonzero] / r[nonzero]

Emag = np.sqrt(Ex**2 + Ey**2)

ux = np.zeros_like(Ex)
uy = np.zeros_like(Ey)

field_nonzero = Emag != 0

ux[field_nonzero] = Ex[field_nonzero] / Emag[field_nonzero]
uy[field_nonzero] = Ey[field_nonzero] / Emag[field_nonzero]

plt.figure(figsize=(8, 7))

plt.contourf(X, Y, V, levels=50)
plt.colorbar(label="Electric Potential (V)")

plt.quiver(
    X[::10, ::10],
    Y[::10, ::10],
    ux[::10, ::10],
    uy[::10, ::10]
)

theta = np.linspace(0, 2*np.pi, 300)

plt.plot(
    R * np.cos(theta),
    R * np.sin(theta),
    linewidth=2,
    label="Sphere boundary"
)

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Electric Potential and Electric Field of a Uniformly Charged Sphere")

plt.legend()

#plt.xlim(-5, 5)
plt.xticks(range(-5, 5, 1))
#plt.ylim(-5, 5)
plt.yticks(range(-5, 5, 1))

plt.gca().set_aspect("equal", adjustable="box")

plt.show()
```

```python
epsilon0 = 8.854e-12
k = 1 / (4 * np.pi * epsilon0)

Q = 1e-9
R = 1.0

x = np.linspace(-5, 5, 201)
y = np.linspace(-5, 5, 201)

X, Y = np.meshgrid(x, y)

r = np.sqrt(X**2 + Y**2)

outside = r >= R
inside = r < R

V = np.empty_like(r)

V[outside] = k * Q / r[outside]

V[inside] = (k * Q / (2 * R)) * (
    3 - r[inside]**2 / R**2
)

dV_dy, dV_dx = np.gradient(V, y, x)

Ex = -dV_dx
Ey = -dV_dy

Emag = np.sqrt(Ex**2 + Ey**2)

ux = np.zeros_like(Ex)
uy = np.zeros_like(Ey)

nonzero = Emag != 0

ux[nonzero] = Ex[nonzero] / Emag[nonzero]
uy[nonzero] = Ey[nonzero] / Emag[nonzero]

plt.figure(figsize=(8, 7))

plt.contourf(X, Y, V, levels=50)
plt.colorbar(label="Electric Potential (V)")

plt.quiver(
    X[::10, ::10],
    Y[::10, ::10],
    ux[::10, ::10],
    uy[::10, ::10]
)

theta = np.linspace(0, 2*np.pi, 300)

plt.plot(
    R * np.cos(theta),
    R * np.sin(theta),
    linewidth=2,
    label="Sphere boundary"
)

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Electric Potential and Electric Field of a Uniformly Charged Sphere")

plt.legend()

plt.xlim(-5, 5)
plt.ylim(-5, 5)

plt.gca().set_aspect("equal", adjustable="box")

plt.show()
```

```python
epsilon0 = 8.854e-12
k = 1 / (4 * np.pi * epsilon0)

Q = 1e-9
R = 1.0

x = np.linspace(-4, 4, 13)
y = np.linspace(-4, 4, 13)
z = np.linspace(-4, 4, 13)

X, Y, Z = np.meshgrid(x, y, z)

r = np.sqrt(X**2 + Y**2 + Z**2)

outside = r >= R
inside = r < R

V = np.empty_like(r)

V[outside] = k * Q / r[outside]
V[inside] = (k * Q / (2 * R)) * (
    3 - r[inside]**2 / R**2
)

E = np.empty_like(r)

E[outside] = k * Q / r[outside]**2
E[inside] = k * Q * r[inside] / R**3

Ex = np.zeros_like(r)
Ey = np.zeros_like(r)
Ez = np.zeros_like(r)

nonzero = r != 0

Ex[nonzero] = E[nonzero] * X[nonzero] / r[nonzero]
Ey[nonzero] = E[nonzero] * Y[nonzero] / r[nonzero]
Ez[nonzero] = E[nonzero] * Z[nonzero] / r[nonzero]

Emag = np.sqrt(Ex**2 + Ey**2 + Ez**2)

ux = np.zeros_like(Ex)
uy = np.zeros_like(Ey)
uz = np.zeros_like(Ez)

field_nonzero = Emag != 0

ux[field_nonzero] = Ex[field_nonzero] / Emag[field_nonzero]
uy[field_nonzero] = Ey[field_nonzero] / Emag[field_nonzero]
uz[field_nonzero] = Ez[field_nonzero] / Emag[field_nonzero]

fig = plt.figure(figsize=(10, 9))
ax = fig.add_subplot(111, projection="3d")

potential = ax.scatter(
    X.flatten(),
    Y.flatten(),
    Z.flatten(),
    c=V.flatten(),
    s=8,
    alpha=0.15
)

ax.quiver(
    X.flatten(),
    Y.flatten(),
    Z.flatten(),
    ux.flatten(),
    uy.flatten(),
    uz.flatten(),
    length=0.35,
    normalize=True
)

u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, np.pi, 50)

Xs = R * np.outer(np.cos(u), np.sin(v))
Ys = R * np.outer(np.sin(u), np.sin(v))
Zs = R * np.outer(np.ones_like(u), np.cos(v))

ax.plot_surface(
    Xs,
    Ys,
    Zs,
    alpha=0.3
)

fig.colorbar(
    potential,
    ax=ax,
    label="Electric Potential (V)",
    shrink=0.6
)

ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.set_zlabel("z (m)")

ax.set_title(
    "3D Electric Potential and Electric Field of a Uniformly Charged Sphere"
)

ax.set_box_aspect([1, 1, 1])

plt.show()
```
